```python
import pandas as pd
import os
import csv
```

```python
def profile_whole_dataset(df):

    profile = {}

    numeric = len(list(df.select_dtypes(include=["int64", "float64"]).columns))
    categoric = len(list(df.select_dtypes(include=["bool", "object"]).columns))

    profile["n_tuples"] = df.shape[0]
    profile["n_attributes"] = df.shape[1]
    profile["p_num_var"] = round(numeric/df.shape[1],2)
    profile["p_cat_var"] = round(categoric/df.shape[1],2)
    profile["p_duplication"] = round(df.duplicated().sum()/df.shape[0],2)

    return profile
```

```python
# Specifica il percorso della cartella
cartella = 'ranking-completeness'

# Lista per salvare i nomi dei file
nomi_file = []

csv_output = 'datasets.csv'

# Scrivi i nomi in un file CSV
with open(csv_output, 'w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f)
    # opzionale: scrivi l'intestazione
    writer.writerow(['name,n_tuples,n_attributes,p_num_var,p_cat_var,p_duplication'])
    for nome in os.listdir(cartella):
        percorso_completo = os.path.join(cartella, nome)
        df = pd.read_csv(percorso_completo)
        profile = profile_whole_dataset(df)
        writer.writerow([nome+','+str(profile["n_tuples"])+','+str(profile["n_attributes"])+','+str(profile["p_num_var"])+','+str(profile["p_cat_var"])+','+str(profile["p_duplication"])])
```
