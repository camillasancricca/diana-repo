```python
d['delta_Q_1'] = 0
d['delta_Q_2'] = 0

for i in range(0, len(d)):
    d['delta_Q_1'].iloc[i] = d.perf_1.values[i] / original_perf[d.algorithm.values[i]]
    d['delta_Q_2'].iloc[i] = d.perf_2.values[i] / original_perf[d.algorithm.values[i]]
```

```python
for m in models:
    d_s['delta_Q_1_'+m] = 0
    d_s['delta_Q_2_'+m] = 0
```

```python
for m in models:
    for i in range(0, len(d_s)):
        d_s['delta_Q_1_'+m].iloc[i] = d_s[m+'_1'].values[i] / original_perf[m]
        d_s['delta_Q_2_'+m].iloc[i] = d_s[m+'_2'].values[i] / original_perf[m]
```

```python
# delta Q nosto approccio (prima dimensione)
for m in models:
    print(m+' MED deltaQ_1 suggested = ' + str(round(d[d['algorithm']==m].delta_Q_1.median(), 4)))
print('\n')
# delta Q tutta la pipeline nostro approaccio (prima+seconda dimensione)
for m in models:
    print(m+' MED deltaQ_2 suggested = ' + str(round(d[d['algorithm']==m].delta_Q_2.median(), 4)))
```
